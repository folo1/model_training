##**4. Custom Training**



1.  The snippet below is currently training with YOLOv8 NANO.
2.  data={dataset.location}/data.yaml should find your .yaml file from the files you extracted in the above snippet. If not, navigate to the left file explorer, find your dataset folder, then right click on data.yaml file and copy-paste 'select path' into here.
3.   Epochs can be set to desired amount of epochs. Be careful of overfitting the greater this number increases.
4.   Imgsize should match the image size you specified in your Roboflow project. Standard sizing for YOLO projects is 640x640.

```python
%cd {HOME}

!yolo task=detect mode=train model=yolov8n.pt data={dataset.location}/data.yaml conf=0.5 epochs=25 imgsz=480,640 plots=True
```
